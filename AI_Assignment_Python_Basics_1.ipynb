```python
from collections import deque

# Missionaries and Cannibals Problem
# State: (M_left, C_left, Boat_position)
# Boat position: 0 = Left bank, 1 = Right bank

INITIAL_STATE = (3, 3, 0)
GOAL_STATE = (0, 0, 1)

# Possible boat actions: (Missionaries, Cannibals)
ACTIONS = [
    (1, 0),  # 1 missionary
    (2, 0),  # 2 missionaries
    (0, 1),  # 1 cannibal
    (0, 2),  # 2 cannibals
    (1, 1)   # 1 missionary and 1 cannibal
]


# Check whether a state is valid
def is_valid(state):
    M_left, C_left, boat = state
    M_right = 3 - M_left
    C_right = 3 - C_left

    if not (0 <= M_left <= 3 and 0 <= C_left <= 3):
        return False

    # Missionaries must not be outnumbered by cannibals
    # on either bank when missionaries are present.
    if M_left > 0 and C_left > M_left:
        return False

    if M_right > 0 and C_right > M_right:
        return False

    return True


# Transition model: apply an action to a state
def transition(state, action):
    M_left, C_left, boat = state
    m, c = action

    if boat == 0:  # Boat travels from left to right
        new_state = (M_left - m, C_left - c, 1)
    else:          # Boat travels from right to left
        new_state = (M_left + m, C_left + c, 0)

    if is_valid(new_state):
        return new_state

    return None


# Generate all possible next states
def get_successors(state):
    successors = []

    for action in ACTIONS:
        new_state = transition(state, action)

        if new_state is not None:
            successors.append((new_state, action))

    return successors


# Goal test
def goal_test(state):
    return state == GOAL_STATE


# Solve the problem using BFS
def solve():
    queue = deque([INITIAL_STATE])
    parent = {INITIAL_STATE: None}
    action_used = {}

    while queue:
        state = queue.popleft()

        if goal_test(state):
            break

        for next_state, action in get_successors(state):
            if next_state not in parent:
                parent[next_state] = state
                action_used[next_state] = action
                queue.append(next_state)

    if GOAL_STATE not in parent:
        print("No solution found.")
        return

    # Reconstruct solution path
    path = []
    state = GOAL_STATE

    while state is not None:
        path.append(state)
        state = parent[state]

    path.reverse()

    # Print solution
    print("Missionaries and Cannibals Solution")
    print("Total crossings:", len(path) - 1)
    print()

    for i, state in enumerate(path):
        M_left, C_left, boat = state
        M_right = 3 - M_left
        C_right = 3 - C_left

        boat_side = "Left" if boat == 0 else "Right"

        print(f"Step {i}:")
        print(f"Left Bank:  Missionaries = {M_left}, Cannibals = {C_left}")
        print(f"Right Bank: Missionaries = {M_right}, Cannibals = {C_right}")
        print(f"Boat Position: {boat_side}")

        if i > 0:
            m, c = action_used[state]
            print(f"Action: Move {m} missionary(s) and {c} cannibal(s)")

        print()


# Run the program
solve()
```
